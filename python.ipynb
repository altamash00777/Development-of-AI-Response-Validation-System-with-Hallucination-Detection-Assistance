```python
# Mutable vs immutable objects
# This is a very important interview topic.
# Mutable	        Immutable
# List       	     Integer
# Dictionary	     String
# Set	             Tuple (its structure)


```

```python
# mutable
numbers = [10, 20, 30]
numbers[0] = 100

print(numbers)
# [100, 20, 30]
```

```python
# immutable
text = "hello"
# text[0] = "H"  # TypeError

text = "H" + text[1:]
print(text)  # Hello
```

```python
# strings
# Important string operations
text = "python programming"

print(len(text))       # 18
print(text.upper())    # PYTHON PROGRAMMING
print(text.lower())    # python programming
print(text.capitalize()) # Python programming
print(text.count("m")) # 2
print(text.find("pro")) # 7
```

```python
# string slicing
text = "Python"

print(text[0:3])  # Pyt
print(text[2:])   # thon
print(text[::-1]) # nohtyP
```

```python
# reverse string
text = "hello"
reversed_text = text[::-1]

print(reversed_text)  # olleh
```

```python
# list
numbers = [10, 20, 30, 20]

numbers.append(40)
numbers.remove(20)

print(numbers)  # [10, 30, 20, 40]
```

```python
# find largest element
numbers = [12, 45, 7, 89, 23]

largest = numbers[0]

for num in numbers:
    if num > largest:
        largest = num

print(largest)  # 89
```

```python
student = {
    "name": "Altamash",
    "age": 21,
    "marks": 85
}

print(student["name"])  # Altamash

student["marks"] = 90
student["city"] = "Lucknow"

print(student)
```

```python
numbers = [1, 2, 2, 3, 1, 2]

frequency = {}

for num in numbers:
    if num in frequency:
        frequency[num] += 1
    else:
        frequency[num] = 1

print(frequency)
# {1: 2, 2: 3, 3: 1}
```

```python
# sets-help in returing unique element
numbers = [1, 2, 2, 3, 3, 4]

unique_numbers = set(numbers)

print(unique_numbers)  # {1, 2, 3, 4}
```

```python
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print(a & b)  # {3, 4}
print(a | b)  # Union of both sets
print(a - b)  # {1, 2}
```

```python
# 1
# Array creation

# np.array(), np.zeros(), np.ones(), np.arange(), np.linspace()

# 2
# Array properties

# shape, ndim, size, dtype

# 3
# Operations

# Indexing, slicing, broadcasting, vectorization, aggregation

# 4
# Array manipulation

# reshape(), flatten(), transpose(), concatenate()
```

```python
# What is the difference between a Python list and a NumPy array?
import numpy as np

python_list = [1, 2, 3]
numpy_array = np.array([1, 2, 3])

print(python_list * 2)  # [1, 2, 3, 1, 2, 3]
print(numpy_array * 2)  # [2 4 6]
```

```python
import numpy as np

arr = np.array([[1, 2, 3], [4, 5, 6]])

print(arr.shape)  # (2, 3)
print(arr.ndim)   # 2
print(arr.size)   # 6
print(arr.dtype)  # An integer data type
```

```python
# vectorization
import numpy as np

arr = np.array([10, 20, 30, 40])

print(arr + 5)
# [15 25 35 45]
```
